**Precision, Recall, and F1-Score**

## 🔹 **1. Precision**

* **Definition:** Measures how many of the **predicted positives are actually correct**.
* Formula:
 Precision = $\frac{TP}{TP + FP}$
* **TP:** True Positive (correctly predicted positive)
* **FP:** False Positive (incorrectly predicted positive)
* **Focus:** “When I say positive, how often am I right?”
* **Use Case:** Important when **false positives are costly**

  * Example: Email spam detection → you don’t want a real email marked as spam

---

## 🔹 **2. Recall (Sensitivity / True Positive Rate)**

* **Definition:** Measures how many of the **actual positives were correctly identified**.
* Formula:
 $$
\text{Recall} = \frac{TP}{TP + FN}
$$

* **FN:** False Negative (missed positive)
* **Focus:** “Of all actual positives, how many did I catch?”
* **Use Case:** Important when **false negatives are costly**

  * Example: Disease detection → you don’t want to miss sick patients

---

## 🔹 **3. F1-Score**

* **Definition:** **Harmonic mean of Precision and Recall**
* Formula:
 $$
F1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
$$

* **Focus:** Balance between Precision and Recall
* **Use Case:** Useful when **both false positives and false negatives matter**

  * Example: Fraud detection → don’t want to miss fraud, but also don’t flag good transactions

---

## 🔹 **Quick Intuition**

* Precision → **how accurate your positive predictions are**
* Recall → **how complete your positive predictions are**
* F1 → **overall balance of Precision & Recall**

---

## 🔹 **Example Table (Confusion Matrix)**

|                 | Predicted Positive | Predicted Negative |
| --------------- | ------------------ | ------------------ |
| Actual Positive | TP = 80            | FN = 20            |
| Actual Negative | FP = 10            | TN = 90            |

* Precision = 80 / (80+10) = 0.89
* Recall = 80 / (80+20) = 0.8
* F1 = 2 * (0.89 * 0.8) / (0.89 + 0.8) ≈ 0.84
